```julia
import pandas as pd
!pip install matplotlib

def clean_data(df):
    # Drop columns: 'Euler_Time', 'Euler_q1' and 14 other columns
    df = df.drop(columns=['Euler_Time', 'Euler_q1', 'Euler_q2', 'Euler_q3', 'Euler_q4', 'Euler_w1', 'Euler_w2', 'Euler_w3', 'SPIRAL_L_Time', 'SPIRAL_q1', 'SPIRAL_q2', 'SPIRAL_q3', 'SPIRAL_q4', 'SPIRAL_w1', 'SPIRAL_w2', 'SPIRAL_w3'])
    return df

# Loaded variable 'df' from URI: Data/analytical_sol.csv
df = pd.read_csv(r'Data/analytical_sol.csv')

df_clean = clean_data(df.copy())
```

```julia
df_clean[df_clean["Rkt_Time"] <= 1.2].plot(x="Rkt_Time", y="Rkt_q1", title="Rocket Quaternions Over Time")
df_clean[df_clean["Rkt_Time"] <= 1.2].plot(x="Rkt_Time", y="Rkt_q2", title="Rocket Quaternions Over Time")
df_clean[df_clean["Rkt_Time"] <= 1.2].plot(x="Rkt_Time", y="Rkt_q3", title="Rocket Quaternions Over Time")
df_clean[df_clean["Rkt_Time"] <= 1.2].plot(x="Rkt_Time", y="Rkt_q4", title="Rocket Quaternions Over Time")
df_clean[df_clean["Rkt_Time"] <= 1.2].plot(x="Rkt_Time", y=["Rkt_q1", "Rkt_q2", "Rkt_q3", "Rkt_q4"], title="Rocket Angular Velocities Over Time")
```
